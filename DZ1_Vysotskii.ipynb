Задача 1
Дан отсортированный по возрастанию массив целых чисел
и некоторое число target
Необходимо найти два числа в массиве, которые в сумме дают
заданное значение target, и вернуть их индексы.

```python
def twoSum(lst:list[int],target:int)->list[int]|bool:
    left=0
    right=len(lst)-1
    while left!=right:
        if lst[left]+lst[right]==target:
            return [left,right]
        elif lst[left]+lst[right]<target:
            left+=1
        else: # lst[left]+lst[right]>target:
            right-=1

    return False
```

Задача решена с помощью 2ух указателей сложность по времени линейная, по памяти константная

Задача 2 Развернуть массив
Дан массив целых чисел.
Необходимо развернуть его.
Сделать это надо за линейное время без дополнительных
аллокаций.

```python
def reverse(lst:list[int],left=0,right=None)->None:
    if right is None:
        right=len(lst)-1
    while left<right:
        lst[left],lst[right]=lst[right],lst[left]
        left+=1
        right-=1
    return # ничего не возвращаем так как решение inplace
```

Задача решена с помощью 2ух указателей сложность по времени линейная, по памяти константная
Добавлена возможность менять границы разворота для следующей задачи

Задача 3 Развернуть часть массива
Дан массив целых чисел.
Необходимо повернуть (сдвинуть) справа налево
часть массива, которая указана вторым параметром.
Сделать это надо за линейное время
без дополнительных аллокаций
Пример:
Исходный массив: 1, 2, 3, 4, 5, 6, 7
➔ k = 3
➔ Результат: 5, 6, 7, 1, 2, 3, 4

```python
def rotate(lst:list[int],k:int)->None:
    k = k%len(lst)
    reverse(lst,left=0,right=len(lst)-k-1)
    reverse(lst,left=len(lst)-k,right=len(lst)-1)
    reverse(lst)
    return # ничего не возвращаем тк решение inplace
# lst=[1,2,3,4,5,6,7]
# rotate(lst,3)
# print(lst)
```

Задча 4 Слияние двух отсортированных массивов
Дано два отсортированных массива.
Необходимо написать функцию которая объединит
эти два массива в один отсортированный.

```python
def mergeSortedArrays(lst1:list,lst2:list)->list:
    res=[]
    first=0
    second=0
    while first<len(lst1) and second<len(lst2):
        if lst1[first]<=lst2[second]:
            res.append(lst1[first])
            first+=1
        else: #lst1[first]>lst2[second]
            res.append(lst2[second])
            second+=1
    else:
        if first==len(lst1):
            while second<len(lst2):
                res.append(lst2[second])
                second+=1
        else: # second==len(lst2)
            while first<len(lst1):
                        res.append(lst1[first])
                        first+=1
    return res
```

Задача решена с помощью двух указателей.
Сложность по времени O(m+n)
Сложность по памяти O(m+n) тк создаем новый массив

создавать третий массив.

```python
def merge(lst1: list[int], m: int, lst2: list[int], n: int) -> None:
    pointer1=m-1
    pointer2=n-1
    resPointer=m+n-1

    while pointer1>-1 and pointer2>-1:
        if lst1[pointer1]>lst2[pointer2]:
            lst1[resPointer]=lst1[pointer1]
            pointer1-=1
            resPointer-=1
        else: # lst1[pointer1]<=lst2[pointer2]:
            lst1[resPointer]=lst2[pointer2]
            pointer2-=1
            resPointer-=1
    # если закончился lst2 ничего не делаем, задача решена
    # закидываем остаток lst2 в lst1

    while pointer2>-1:
            lst1[resPointer]=lst2[pointer2]
            pointer2-=1
            resPointer-=1

    return # ничего не возвращаем тк решение inplace
    



```

Задача 6 Минимальный размер подмассива
Дан массив положительных целых чисел nums и
положительное целое число target. Верните
подмассив минимальной длины, сумма элементов
которого больше или равна target. Если такого
подмассива не существует, верните 0.

```python
def minSubArray(nums:list[int], target:int)->list[int]:

    s=0
    res=[0,float("inf")]

    fast=0
    slow=0

    while fast<len(nums):
        s+=nums[fast]
        if s<target:
            fast+=1
        else:
            while s>=target:
                if fast-slow<res[1]-res[0]:
                    res=[slow,fast]
                s-=nums[slow]
                slow+=1

    return nums[res[0]:res[1]+1] if res[1]!=float("inf") else 0

```

Задача решена с помощью двух указателей и sliding window
Временная сложность O(n)
Сложность по памяти O(1)  

Задача 7 Сортировка массива из 0 и 1
Дан массив, содержащий только 0 и 1.
Напишите функцию, которая сортирует его так, чтобы все нули оказались в начале, а все единички — в конце.
Решение должно быть in-place.

```python
def sortBinaryArray(lst:list[int])->None:

    left=0
    right=len(lst)-1

    while left<right:
        if lst[left]==0:
            left+=1
        elif lst[right]==1:
            right-=1
        else: # left на единице и right на 0
            lst[left],lst[right]=lst[right],lst[left]
            left+=1
            right-=1

    return # Ничего не возвращаем тк решение in place
```

Задача решена с помощью двух указателей
Сложность по времени O(n)
Сложность по памяти константная

Задача 8 Флага Нидерландов
Дан массив состоящий из нулей,
единиц и двоек
Необходимо его отсортировать
за линейное время

```python
def sortColors(nums: list[int])->None:

    left=0
    right=len(nums)-1
    pointer=0

    while pointer<=right:
        if nums[pointer]==0:
            nums[pointer],nums[left]=nums[left],nums[pointer]
            left+=1
            pointer+=1
        elif nums[pointer]==1:
            pointer+=1
        else: # nums[pointer]==2
            nums[pointer],nums[right]=nums[right],nums[pointer]
            right-=1

    return # Ничего не возвращаем тк решение in place
```

Задача решена с помощью 3х указателей
Сложность по времени линейная
Сложность по памяти константная

Задача 9 Передвинуть четные числа вперед
Дан не отсортированный массив целых чисел
Необходимо перенести в начало массива все
четные числа, сохраняя их очередность
То есть если 8 стояла после 2, то после
переноса в начало, она по-прежнему должна
стоять после 2
Пример ввода: [3, 2, 4, 1, 11, 8, 9]
Пример вывода: [2, 4, 8, 1, 11, 3, 9]

```python
def evenFirst(lst:list[int])->list[int]:
    evenPointer=0

    for i in range(len(lst)):
        if lst[i]%2==0:
            lst[i],lst[evenPointer]=lst[evenPointer],lst[i]
            evenPointer+=1
            
    return lst
```

Задача решена с помощью 2х указателей
Сложность по времени линейная
Сложность по памяти константная

Задача 10 Нули в конец
В школе прошел экзамен по математике.
Несколько человек списали решения
и были замечены. Этим школьникам
поставил 0 баллов.
Задача: есть массив с оценками, среди
которых есть 0. Необходимо все оценки,
равные нулю перенести в конец массива,
чтобы все такие школьники оказались
в конце списка

Пример ввода: [0, 0, 1, 0, 3, 12]
Пример вывода: [1, 3, 12, 0, 0, 0]

```python
def zerosToEnd(lst:list[int])->list[int]:
    slow=0
    fast=0
    while fast<len(lst):
        if lst[slow]!=0:
            slow+=1
        elif lst[fast]==0:
            fast+=1
        else:
            lst[slow],lst[fast]=lst[fast],lst[slow]
            
    return lst
```

Задача решена с помощью 2х указателей
Сложность по времени линейная
Сложность по памяти константная
