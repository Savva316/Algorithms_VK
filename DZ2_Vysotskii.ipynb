```python
# Далее для связного списка будем использовать этот класс
class ListNode:
    def __init__(self, x):
        self.val = x
        self.next = None
```

Задача 1
Проверить, является ли список циклическим
Дан односвязный список.
Необходимо проверить, является ли этот список
циклическим.

```python

def hasCycle(head:ListNode|None) -> bool:
    if not head:
            return False
    slow = head
    fast = head

    while fast:
        slow=slow.next
        fast=fast.next
        if fast:
            fast=fast.next
        else:
            break
        if not fast:
            return False

        if slow==fast:
            return True # циклический

    return False # не циклический
```

```python
# циклический:
first=ListNode(1)
second=ListNode(2)
third=ListNode(3)
fourth=ListNode(4)

first.next=second
second.next=third
third.next=fourth
fourth.next=second

print(hasCycle(first))
```

```python
# не циклический
first=ListNode(1)
second=ListNode(2)
third=ListNode(3)
fourth=ListNode(4)

first.next=second
second.next=third
third.next=fourth
fourth.next=None

print(hasCycle(first))
```

Задача решена с помощью 2ух указателей сложность по времени линейная, по памяти константная

Задача 2 Развернуть односвязный список
Необходимо написать функцию, которая
принимает на вход односвязный список
и разворачивает его.

```python
def reverse(head:ListNode|None) -> ListNode|None:
    if not head:
        return head #нечего разворачивать

    prev=None
    pointer=head

    while pointer:
        right=pointer.next
        pointer.next=prev
        prev=pointer
        pointer=right

    return prev
```

Задача решена с помощью 2ух указателей сложность по времени линейная, по памяти константная

Задача 3 Найти середину списка
Дан связный список.
Необходимо найти середину списка.
Сделать это необходимо за O(n)
без дополнительных аллокаций

```python
def middleNode(head:ListNode|None) -> ListNode|None:
    if not head:
        return head 
    
    fast=head
    slow=head

    while fast:
        fast=fast.next
        if fast:
            fast=fast.next
        else:
            return slow
        slow=slow.next
        
    return slow

    
```

Задача 4 Удалить элемент из односвязного списка
Необходимо написать функцию, которая
принимает на вход односвязный список
и некоторое целое число value. Необходимо
удалить узел из списка, значение которого
равно value.

```python
def removeElements(head:ListNode|None, value:int) ->ListNode|None:

    if not head:
        return head
    
    dummy=ListNode(0)
    dummy.next=head

    prev=dummy
    pointer=head
    while pointer:

        if pointer.val==value:
            nxt=pointer.next
            prev.next=nxt
            pointer=nxt
        else:
            pointer=pointer.next
            prev=prev.next

    return dummy.next
```

Сложность по времени линейная
Сложность по памяти константная

Задача 5 Удаление n-ого элемента с конца списка
Дан head односвязного списка и целое число n.
Необходимо удалить n-й узел с конца и вернуть
обновлённый список. Например, если вход -
[1,2,3,4,5] и n = 2, результатом будет [1,2,3,5].

```python
def removeNthFromEnd(head: ListNode | None, n: int) -> ListNode | None:
    dummy=ListNode(0)
    dummy.next=head

    fast=dummy
    slow=dummy
    for i in range(n+1):
        fast=fast.next
    while fast:
        slow=slow.next
        fast=fast.next

    slow.next=slow.next.next

    return dummy.next
```

Задача решена с помощью 2ух указателей.
Сложность по времени линейная
По памаяти константная

Задача 6 Является ли одна строка исходной для другой
В исходную строку добавили некоторое
количество символов.
Необходимо выявить, была ли строка
a исходной для строки b.

```python
def isSubsequence(a: str, b: str)-> bool:

    if not a:
        return True # если а пустая то она является подстрокой для луюой строки

    if len(a)>len(b):
        return False #невозможно чтобы подстрока была длиннее строки 
    
    pointer=0
    for letter in b:
        if a[pointer]==letter:
            pointer+=1
            if pointer==len(a):
                return True
    return False


```

```python

print(isSubsequence("abba","fafbbhja"))
```

Задача решена с помощью двух указателей 
Временная сложность O(len(b)) - то есть линейная 
Сложность по памяти O(1)  

Задача 7 Является ли слово палиндромом?
Напишите функцию, которая принимает
на вход строку и возвращает true, если
она является палиндромом*.
В противном случае верните false.
*слово или текст, одинаково читающиеся
в обоих направлениях.

```python
def isPalindrome(s : str) -> bool:

    left=0
    right=len(s)-1

    while left<right:
        if s[left]==s[right]:
            left+=1
            right-=1
        else:
            return False

    return True
```

```python
print(isPalindrome("abba"))
```

Задача решена с помощью двух указателей
Сложность по времени O(n)
Сложность по памяти константная

Задача 8 Сортировка стека через дополнительный стек

```python
def sortArrayUsingStack (initial_stack: list[int]) -> list[int]:
    additional_stack=[]

    while initial_stack:
        el=initial_stack.pop()
        while additional_stack and additional_stack[len(additional_stack)-1]>el:
            tmp=additional_stack.pop()
            initial_stack.append(tmp)

        additional_stack.append(el)
    
    return additional_stack
```

```python
res=sortArrayUsingStack([2,3,1,4,5,6,9,2,3,1,5,6,8,3,7,9,0])
print(res)
```

Задача решена с помощью аллокации доп памяти
Сложность по времени O(n^2)
Сложность по памяти линейная

Задача 9 Слияние двух отсортированных списков
Написать функцию, которая принимает на вход
два отсортированных односвязных списка и
объединяет их в один отсортированный список.
При этом затраты по памяти должны быть O(1)
Входные данные: list1 = [3, 6, 8], list2 = [4, 7, 9, 11]
Выходные: [3,4,6,7,8,9,11]

```python
def mergeSortedList(l1: ListNode,l2:ListNode)->ListNode:
    # l1 и l2 указатели на головы списков 1 и 2 соответственно
    dummy=ListNode(0)
    dummy.next=l1
    prev=dummy
    while l1 and l2:
        if l1.val<l2.val:
            l1=l1.next
            prev=prev.next
        else: #l1.val>=l2.val:
            nxt=l2.next

            prev.next=l2
            prev=prev.next

            l2.next=l1
            l2=nxt
    # если закончился l1
    while l2:
        prev.next=l2
        l2=l2.next
        prev=prev.next
    # если закончился l2
    # все ОК, все элементы перешли в l1, ничего не делаем
    

    return dummy.next
```

```python
#l1
first1=ListNode(3)
second1=ListNode(6)
third1=ListNode(8)


first1.next=second1
second1.next=third1
third1.next=None

#l2
first2=ListNode(4)
second2=ListNode(7)
third2=ListNode(9)
fourth2=ListNode(11)

first2.next=second2
second2.next=third2
third2.next=fourth2
fourth2.next=None

head=mergeSortedList(first1,first2)
while head:
    print(head.val,end=" ")
    head=head.next
```
