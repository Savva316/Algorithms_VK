Задача 1. Найти корень числа (ближайшее целое)
Необходимо написать функцию, которая находит целую часть квадратного корня числа
(округление вниз).
Что нужно сделать:
Реализуйте решение с использованием бинарного поиска без использования
встроенных функций извлечения корня. Объясните, почему алгоритм работает за O(log
n) и как определяется граница поиска.

```python
def binarySearchSqrt(target) -> int:
    if target<2:
        return target
    
    ans=left=0 # нижняя граница тк корень это неотрицательное число 
    right=target/2 # для x>=2 это верхняя граница корня
    while left<=right:
        mid=(right+left)//2
        if mid*mid==target:
            return mid
        elif mid*mid<target:
            ans=mid # сохраняем возможный ответ если вдуг завершим цикл while ( это будет округление вниз)
            left=mid+1
        else:
            right=mid-1
            
    return ans
    
```

```python
print(binarySearchSqrt(26))
```

Задача решена  на основе бинарного поиска, сложность которого O(log n) 

Задача 2 Очень лёгкая задача (два ксерокса)
Как быстро можно сделать N копий документа, если есть два ксерокса, каждый
копирует с разной скоростью — x и y минут за копию?
Что нужно сделать:
Реализуйте алгоритм нахождения минимального времени, за которое будет сделано N
копий. Используйте бинарный поиск по времени и поясните, как проверяется условие
достаточности времени.

```python
# если оба принтера начинают делать копии сразу вместе
def copyTimeByTwoCopiers(n:int, x:int, y:int) -> int:
    left=0 # самае маленькое время это когда у нас 0 копий, а значит и времени не надо
    right=n*max(x,y) # самое большое время это когда самый долгий принтер делает все копии
    while left+1<right:
        mid=(left+right)//2
        copies=mid//x + mid//y # количество копий за время mid двумя принтерами
        if copies<n:
            left=mid
        else:
            right=mid

    copies_left=left//x+left//y
    if copies_left>=n:
        return left
    else:
        return right
# здесь была использована идея первого вхождения слева, что соответствует минимальному времени
```

Но у нас же сначала нужно сделать копию первую, а потом запустить в работу два принтера вместе, тогда итоговый результат будет выдавать функция ниже

```python
def copyTime(n:int, x:int, y:int) -> int:
    if n == 1:
        return min(x, y) # второй принтер не нужен
    return min(x,y)+copyTimeByTwoCopiers(n-1,x,y) # сделали копию на самом быстром принтере и потом запустили копирование на двух принтерах
```

```python
print(copyTime(5,2,3))
```

Задача решена с помощью левого бинарного поиска, сложность по времень логорифмическая, по памяти константная

Задача 3 Накормить животных
В зоопарке есть животные, каждое из которых требует определённое количество еды
(целое число).
Каждая привезённая порция еды может накормить только одно животное и не
делится.
Необходимо определить, сколько животных можно накормить имеющимся набором
порций.
Что нужно сделать:
Реализуйте жадный алгоритм распределения еды (например, сортировка животных и
порций). Объясните, почему выбранная стратегия даёт максимальное количество
накормленных животных.

Сначала реализуем какую-нибудь сортировку, потому что для этой задачи нам понадобится она, будем реализовывать быструю сортировку

```python
def quickSort(nums:list[int])->list[int]:
    def quickSortRecursive(low:int,high:int):
        if low>=high:
            return
        left=low
        right=high
        pivot=nums[(left+right)//2]
        while left<=right:
            while nums[left]<pivot:
                left+=1
            while nums[right]>pivot:
                right-=1
            if left<=right:
                nums[left],nums[right]=nums[right],nums[left]
                left+=1
                right-=1
    
        quickSortRecursive(0,right)
        
        quickSortRecursive(left,high)

    quickSortRecursive(0, len(nums) - 1)
    return nums


    
```

```python
print(quickSort([1,4,6,3,5,6,8,9,0,2,4,6,9]))
```

```python
def feedAnimals(animals:list[int], food:list[int]) -> int:
    quickSort(animals)
    quickSort(food)
    count=0
    for portion in food:
        if count>len(animals)-1:
            break
        if portion>=animals[count]:
            count+=1

        
    return count

    
```

```python
print(feedAnimals([1, 1, 1],[1, 2, 5,3]))
```

Когда мы сортируем порции еды по возрастанию, мы заставляем алгоритм брать самый маленький мешок из всех возможных, способных насытить конкретное животное. Сортируя животных по возрастанию аппетита, мы пытаемся в первую очередь накормить тех, кому нужно меньше всего еды.

Задача решена с помощью жадного алгоритма, сложность по времени O(nlog n + mlog m ) так как сортируем два массива,сложность по памяти 
O(log n + log m ) так как мы использовали рекурсивную быструю сортировку.

Задача 4 Найти разницу между двумя строками
Даны строки a и b.
Строка b получена из строки a путём перемешивания символов и добавления одной
новой буквы.
Необходимо вернуть эту добавленную букву.
Что нужно сделать:
Реализуйте решение без сортировки строк (желательно за O(n)). Объясните, как
используется подсчёт символов или XOR для нахождения лишнего элемента

Ниже предложены две реализации, через XOR и через счетчик словарь


Для XOR a^a=0 поэтому одинаковые символы уничтожаются и остается один у которого не нашлось пары в другом слове

```python
def extraLetterXOR(a: str, b: str)->str:
    res=ord(a[0])
    for i in range(1,len(a)):
        res^=ord(a[i])

    for char in b:
        res^=ord(char)
    return chr(res)
```

```python
print(extraLetterXOR("sdfaefwgfsdgfsd","sdfaefwLgfsdgfsd"))
```

Сложность по времени линейная от числа входных символов, по памяти константная (для extraLetterXOR)

Для словаря мы добавляем все символы из а как ключ и присваеваем им значение 1, потом проходим по символам b и вычитаем в словаре 1 , в итоге у нас останутся все ключи со значениями 0 и один с минус единичкой - это тот который нам нужен (my_dict=defaultdict(int) создает словарь со значениями int по умолчанию равными 0, поэтому все ОК когда прибавляем или вычитаем из "необъявленных" значений, они объявлены 0 по умолчанию)

```python
from collections import defaultdict
def extraLetterHash(a: str, b: str)->str:
    my_dict=defaultdict(int)
    for char in a:
        my_dict[char]+=1 # тут все ОК тк my_dict[char] по умолчанию равен 0 если его еще не объявляли

    for char in b:
        my_dict[char]-=1 # тут все ОК
    
    for key in my_dict:
        if my_dict[key]==-1:
            return key
```

```python
print(extraLetterHash("sdfaefwgfsdgfsd","sdfaefwLgfsdgfsd"))
```

Сложность по времени линейная от числа входных символов, по памяти O(len(b)) (для extraLetterHash)

Задача 5 Сумма двух элементов массива
Дан неотсортированный массив целых чисел и число target.
Необходимо найти два элемента массива, сумма которых равна target.
Каждый элемент можно использовать только один раз.
Если подходящей пары нет — вернуть пустой массив.
Что нужно сделать:
Реализуйте эффективное решение за O(n), используя хеш-таблицу (словарь).
Поясните, как хранение уже просмотренных элементов позволяет избежать двойного
цикла.

Это самая первая задача с литкода, можно решить брут форс подходом за n^2 времени , но существует эффективное решение за линейную по памяти и времени сложность с помощью хэш таблицы,
будем идти по списку и искать нужное нам второе число как x=target-lst[i], если мы встречали число x то оно будет в словаре, и мы нашли ответ, если же x не в словаре  то добавляем lst[i] в словарь и идем далее,

```python
def twoSum(lst: list[int],target: int) -> list:
    my_dict=dict()
    for i in range(len(lst)):
        x=target-lst[i]
        if x not in my_dict:
            my_dict[lst[i]]=i
        else:
            return [i,my_dict[x]] # возвращаем индексы двух элементов начальном списке
    return []

```

```python
print(twoSum([3,3],6))
```

Задача 6 Массив анаграмм
Дан массив строк strs. Сгруппируйте анаграммы вместе. Вы можете вернуть ответ в
любом порядке.
Анаграмма — это слово или фраза, полученная путем перестановки букв другого
слова или фразы, обычно с использованием всех исходных букв ровно один раз.
Что нужно сделать:
Пример: ["eat","tea","tan","ate","nat","bat"] -> [["bat"],["nat","tan"],["ate","eat","tea"]]
Алгоритм нужно выбрать самим и стремиться сделать наиболее эффективно

```python
from collections import defaultdict
def groupAnnagrams(words: list[str], )-> list[list[str]]:
    res=[]
    len_alphabet=ord("z")-ord("a")+1
    ord_a=ord("a")
    my_dict=defaultdict(list)
    for word in words:
        key_alphabet=[0]*len_alphabet
        for char in word:
            idx=ord(char)%ord_a
            key_alphabet[idx]+=1
        my_dict[tuple(key_alphabet)].append(word)

    for key in my_dict:
        res.append(my_dict[key])

    
    return res
```

```python
print(groupAnnagrams(["eat","tea","tan","ate","nat","bat"]))
```

Сложность по времени O(n*m), где n - количество входных слов, m - средняя длина слова (можно сказать что сложность линейная по количеству входных символов)
Сложность по памяти O(2*n*m+константное значение на английский алфивит и подобное)=O(n*m) так как мы используем defultdict где по сути храним весь вход, потом еще перезаписываем все в res, еще для каждого слова храним массив длиной английского алфавита
