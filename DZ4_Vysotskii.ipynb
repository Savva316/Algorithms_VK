В качестве узлов дерева будет использоваться следующая структура:

```python
class Node():
    def __init__(self,val=0,left=None,right=None):
        self.value = val
        self.left = left
        self.right = right
```

1. Восстановление бинарного дерева из массива
Дан массив, представляющий бинарное дерево (например, в виде уровнями —
level-order).
Необходимо восстановить структуру бинарного дерева.
Что нужно сделать:
Реализуйте функцию, которая строит дерево по переданному массиву. Объясните, как
определяется связь родитель–потомок и какая структура данных используется для
хранения узлов.

```python
def buildTree(lst: list[int],i) -> Node:
    if i>=len(lst):
        return None
    root=Node(lst[i])
    root.left=buildTree(lst,2*i+1)
    root.right=buildTree(lst,2*i+2)

    return root
```

левый потомок родителя с индексом i это индекс 2i+1 если он в пределах списка, если не в пределах значит это был лист
правый потомок родителя с индексом i это индекс 2i+2 если он в пределах списка, если не в пределах значит это был лист

2. Симметричное бинарное дерево
Дано бинарное дерево.
Необходимо определить, является ли оно симметричным относительно своей
середины.
Что нужно сделать:
Реализуйте рекурсивную или итеративную проверку зеркальности левого и правого
поддеревьев. Поясните, какие пары узлов сравниваются и почему алгоритм работает
корректно.

```python

# обойдем дерево в ширину и будем проверять является ли каждый уровень палиндромом, если является то дерево симметричное
from collections import deque
def isPalindrome(lst: deque[Node])-> bool:
    left=0
    right=len(lst)-1
    while left<right:

        a = lst[left]
        b = lst[right]
        
        if a is None and b is None:
            pass
        elif a is None or b is None:
            return False
        elif a.value != b.value:
            return False
    return True


def isSymmetric(root: Node|None)->bool:
    if not root:
        return True # Считаем пустое дерево симметричным
    
    deq=deque([root])
    while deq:
        if not isPalindrome(deq):
            return False
        for _ in range(len(deq)):
            node=deq.popleft()
            if node is not None:
                deq.append(node.left)
                deq.append(node.right)
    return True
```

```python
# бинарное дерево
root = Node(1)

root.left = Node(2)
root.right = Node(2)

root.left.left = Node(3)
root.left.right = Node(4)

root.right.left = Node(4)
root.right.right = Node(3)
```

```python
print(isSymmetric(root))
```

3. Поиск минимальной глубины бинарного дерева
Дано бинарное дерево.
Необходимо найти его минимальную глубину (расстояние от корня до ближайшего
листа).
Что нужно сделать:
Реализуйте решение с использованием BFS или DFS. Объясните, почему поиск в
ширину (BFS) позволяет корректно найти минимальную глубину.

```python
def minHeight(root:Node|None)->int:
    if not root:
        return 0 # дерево пустое -> высота 0
    deq=deque([root])
    h=1 # так как глубина определяется количеством узлов на пути от корня до ближайшего листа, то инициализируем 1
    while deq:
        for _ in range(len(deq)):
            node=deq.popleft()
            if not node.left and not node.right:
                return h 
            if node.left:
                deq.append(node.left)
            
            if node.right:
                deq.append(node.right)
        h+=1
```

BFS обходит дерево уровень за уровнем. Поэтому первый найденный лист обязательно находится на минимальной глубине.

4. Поиск произведения максимального и минимального элементов
Дано бинарное дерево поиска (BST), представленное в виде массива.
Необходимо найти произведение минимального и максимального значений.
Что нужно сделать:
Восстановите BST (если требуется) и найдите минимальный и максимальный
элементы, используя свойства BST. Поясните, почему поиск крайних элементов
выполняется за O(h), где h — высота дерева.

```python
def minMaxProduct(root:Node|None)->int:
    def minVal(root):
        if not root:
            return None
        
        if root.left:
            return minVal(root.left)
        else:
            return root.value
    def maxVal(root):
        if not root:
            return None
        
        if root.right:
            return maxVal(root.right)
        else:
            return root.value
        
    a=minVal(root)
    b=maxVal(root)

    return a*b
```

```python
# бинарное дерево
root = Node(8)

root.left = Node(3)
root.right = Node(10)

root.left.left = Node(1)
root.left.right = Node(6)

root.left.right.left = Node(4)
root.left.right.right = Node(7)

root.right.right = Node(14)
root.right.right.left = Node(13)
```

```python
print(minMaxProduct(root))
```

мы не перебираем все n узлов дерева, а проходим только один путь от корня вниз. Именно поэтому сложность зависит от высоты h, а не напрямую от количества узлов n.

5. Сравнение двух бинарных деревьев
Даны два бинарных дерева.
Необходимо определить, являются ли они идентичными (по структуре и значениям
узлов).
Что нужно сделать:
Реализуйте рекурсивную проверку соответствия узлов обоих деревьев. Объясните,
какие условия должны выполняться, чтобы деревья считались одинаковыми.

```python
def isSameTree(root_a:Node|None,root_b:Node|None)->bool:
    if not root_a and not root_b:
        return True
    if  not root_a and root_b:
        return False
    if root_a and not root_b:
        return False
    if root_a.value!=root_b.value:
        return False

    return isSameTree(root_a.left, root_b.left) and isSameTree(root_a.right, root_b.right)
```

Оба узла равны None — возвращаем True.

Один узел равен None, а другой нет — возвращаем False.

Значения узлов различаются — возвращаем False.

Значения совпадают — рекурсивно сравниваем левые поддеревья между собой и правые между собой.
