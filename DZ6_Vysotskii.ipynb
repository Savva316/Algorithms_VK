1. Наибольшая непрерывная возрастающая последовательность
Дан массив не отсортированных чисел. Необходимо найти максимально длинную
непрерывную возрастающую последовательность и вернуть её длину.
Примеры:
[3, 2, 8, 9, 5, 10] — ответ 3, так как максимально длинная последовательность: 2, 8, 9
[1, 2, 7, 9, 0, 10] — ответ 4, так как максимально длинная последовательность: 1, 2, 7,
9
[8, 8, 8, 8] — ответ 1
Проанализируйте решение за O(n) и объясните, как отслеживается текущая и
максимальная длина последовательности.

```python
def findLIS(nums:list[int]) -> int:
    if not nums:
        return 0
    elif len(nums)==1:
        return 1
    largest=1
    current=1
    for i in range(1,len(nums)):
        if nums[i]>nums[i-1]:
            current+=1
        else:
            largest=max(largest,current)
            current=1
    return max(largest, current) # вдруг самая длинная последовательность находится в конце цикла

print(findLIS([1, 2, 7, 9, 0, 10]))
```

Сложность по памяти константная, по времени линейная
Мы идем по массиву и считаем длину каждой подходящей последовательности, если она больше предыдущей подходящей, то перезаписываем ее как самую длинную, так делается для каждой подходящей последовательности

2. Треугольник Паскаля
Дано число n. Необходимо построить треугольник Паскаля из n строк.
Объясните, как формируется каждая строка на основе предыдущей и какова
сложность алгоритма.

```python
def pascalTraingle(n:int)-> list[list[int]]:
    
    if n==1:
        return [[1]]
    res=[[1]]
    for i in range(1,n):
        lst=[]
        for j in range(len(res[i-1])+1):
            if j==0:
                lst.append(1)
                continue
            if j==len(res[i-1]):
                lst.append(1)
                continue
            num=res[i-1][j-1]+res[i-1][j]
            lst.append(num)
        res.append(lst)
    return res

print(pascalTraingle(4))
```

Основная часть алгоритма находится во втором for , то есть если элементы краевые то они равны единичке, если нет (пусть рассматриваем элемент с индексом j) то это сумма элементов с прошлой строки под индексами j и j-1 для j ого элемента текущей строки
Сложность по времени O(1+2+3+...+n)=O(n(n+1)/2)=O(n^2)
Сложность по памяти O(n^2) тк нужно хранить все элементы треугольника

3. Размен монет
Дан массив номиналов монет coins и число amount. Требуется найти минимальное
количество монет, составляющих сумму amount, либо вернуть -1, если это
невозможно.
Реализуйте решение с использованием динамического программирования и
обоснуйте корректность переходов.

```python
def coinChange(coins:list[int],amount:int)->int:
    res=[-1]*(amount+1)
    res[0]=0
    for i in range(1,len(res)):
        for coin in coins:
            if i-coin>=0 and res[i-coin]!=-1:
                if res[i]==-1:
                    res[i]=res[i-coin]+1
                else:
                    res[i]=min(res[i],res[i-coin]+1)
    return res[-1]

coins = [1,2,5]
amount = 11
print(coinChange(coins,amount))
```

len(amount)=m,len(coins)=n
Сложность по времени O(nm)
Сложность по памяти  O(m)
Для каждой суммы мы проверяем можно ли в нее дойти с каждым номиналом монеты и выбираем минимальное число монет которое нужно чтобы попасть в эту сумму

4. Максимальный палиндром в строке
Дана строка s. Необходимо найти её самую длинную подстроку-палиндром.
Рассмотрите подход с динамическим программированием или расширением от центра
и проанализируйте сложность алгоритма.

```python
def longestPalindrome(s: str) -> int:
    def expand(left, right):
        while left >= 0 and right < len(s):
            if s[left] == s[right]:
                left -= 1
                right += 1
            else:
                break

        return left+1,right-1

    start, end = 0, -1
    for i in range(len(s)):
        l1, r1 = expand(i, i)
        l2, r2 = expand(i, i+1)

        if r1 - l1 > end - start:
            start, end = l1, r1

        if r2 - l2 > end - start:
            start, end = l2, r2
    return s[start:end+1]
print(longestPalindrome("babad"))
```

Сложность по времени O(n^2) тк для каждого символа строки в худшем случае необходимо n сравнений
Сложность по памяти  O(1) тк мы используем только указатели на начало и конец строки и на каждой итерации обновляем их же
